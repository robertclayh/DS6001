```python
import numpy as np
import pandas as pd
import requests
import json
import dotenv
import os
```

A "dot E N V" file, named .env, is a plain text file for storing passwords/keys/secrets.

Because it begins with a . for a filename, it is understood by your computer to be a hidden file. That's extra security.

When using GitHub, always use a .gitignore file. .env will always be one of the ignored files. DO NOT allow your .env file to be uploaded to GitHub.

To load secret keys from the .env file:

```python
dotenv.load_dotenv()
newskey = os.getenv('newskey')
```

```python
url = 'https://newsapi.org/v2/everything'
params = {'apiKey':newskey,
          'q':'Claudia Sheinbaum'}
r = requests.get(url, params=params)
r
```

```python
myjson = r.json()
news_df = pd.json_normalize(myjson, record_path=['articles'])
news_df
```

## Congress API

```python
congresskey = os.getenv('congresskey')
```

### User agent strings

A user agent string is a string that informs the API what kind of system you are using to access the API. It's considered good etiquette to include one, and it is sometimes required by the API.

https://httpbin.org/user-agent

```python
url = 'https://httpbin.org/user-agent'
r = requests.get(url)
myjson = r.json()
useragent = myjson['user-agent']
```

```python
headers = {'User-Agent': useragent,
           'From': 'robertclayh@icloud.com'}
```

Four things to look for in API documentation:

1. The root
2. The endpoint for the data source you need
3. The parameters
4. Headers, or other instructions for supplying your API key

```python
root = 'https://api.congress.gov/v3' # the beginning of all URLs attached to this API
endpoint = '/member' # the endpoint we want to access
params = {'format':'json',
          'limit':250,
          'offset':0,
          'currentMember':'true',
          'api_key':congresskey} # the parameters we want to pass to the API
r = requests.get(root+endpoint, headers=headers, params=params) # the full URL we want to access
r
```

```python
myjson = json.loads(r.text)
```

```python
congress_df = pd.json_normalize(myjson, record_path=['members'])
congress_df
```

```python
dotenv.load_dotenv()
feckey = os.getenv('feckey')
if feckey is None:
    print("API key not found. Please set the 'feckey' environment variable.")
else:
    print("API key found.")
```

Four things to look for in API documentation:

1. The root
2. The endpoint for the data source you need
3. The parameters
4. Headers, or other instructions for supplying your API key

```python
# Define the root URL and endpoint
root = 'https://api.open.fec.gov'
endpoint = '/v1/candidates/search'

# Define the parameters, including the API key
params = {
    'api_key': feckey,
    'q': 'Trump'
}

# Make the GET request
r = requests.get(root + endpoint, params=params, headers=headers)

# Check the response
print(r)
print(r.json())
```
